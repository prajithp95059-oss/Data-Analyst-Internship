<h2 style="color:teal;">Observations</h2>

- Peak Sales in September: September recorded the highest burger sales with 104 orders, suggesting strong demand during this month—possibly due to seasonal promotions, weather, or consumer behavior.
- Consistent Mid-Year Performance: August and October both had 101 orders, showing stable performance around the late monsoon and early festive season.
- Early Year Stability: January and February each had 96 orders, indicating a steady start to the year without major fluctuations.
- Lowest Sales in November: November saw the lowest burger sales at 84 orders, which might reflect shifting consumer preferences, competing food trends, or reduced marketing efforts
